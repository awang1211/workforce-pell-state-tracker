# Build the state panel
One row per state: implementation stage (frozen 2026-11-17) + capacity + need.

```python
import pandas as pd

# Frozen snapshot (stages verified through 2026-11-17)
status   = pd.read_csv("../data/state_status_2026-11-17.csv")
capacity = pd.read_csv("../data/capacity_hcm2024.csv")
need     = pd.read_csv("../data/need_lumina_2024.csv")

need_small = need[["State_Code", "No_Credential_25_64"]].copy()

# Merge: tracker + capacity + need (left join keeps all 50 states)
panel = status.merge(capacity, on="State_Code", how="left")
panel = panel.merge(need_small, on="State_Code", how="left")
print(panel.shape)  # expect 50 rows

panel["Stage_Final"] = panel["Stage_Order"]
```

```python
# Quality checks
print(panel["Stage_Final"].value_counts().sort_index())   # expect 0:3, 1:8, 3:28, 4:11
print(panel[["Stage_Final", "student_aid", "No_Credential_25_64"]].isna().sum())  # expect all 0

# Save the analysis dataset
panel.to_csv("../data/state_panel.csv", index=False)
```
